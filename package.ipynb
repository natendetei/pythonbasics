# Step 2: Create a employee data set

```python
datas = {
    'Name':["Mary","Otieno","James","Mercy","Lucy"],
    'Department':["IT","HR","Finance","IT","Sales"],
    'Age':[25,30,28,26,32],
    'Salary':[45000,35000,28000,50000,60000]
}
```

# Step 3: Create a dataframe
# Step 4: Display all employees

```python
employee_data = pd.DataFrame(datas)
print(employee_data)
```

# Display employees names only

```python
print(employee_data.Name)
```

# Display employees earning more than 30000

```python

print(employee_data[employee_data["Salary"]>=30000])
```

# Employee with highest salary

```python
print(employee_data["Salary"].max())
```

# Mean salary

```python
print(employee_data["Salary"].mean())
```
