# Introduction to matplotlib
Matplotlib is a python library used to create graphs and charts from data

```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
datas = {
    'Name':["Mary","Otieno","James","Mercy","Lucy"],
    'Salary':[45000,35000,28000,50000,60000]
}
employees = pd.DataFrame(datas)
plt.bar(employees["Name"],employees["Salary"])
plt.title("employees salaries")
plt.xlabel("Employee Name")
plt.ylabel("Employee Salary")
```

# Creating a line graph

```python
months = ["January","February","March","April","May"]
sales = [10000,15000,12000,20000,25000]
plt.plot(months,sales)
plt.title("Monthly Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
```
