```python
import matplotlib.pyplot as plt
```

```python
import pandas as pd
datas = {
    'Name':["Mary","Otieno","James","Mercy","Lucy"],
    'Salary':[45000,35000,28000,50000,60000]
}
employees = pd.DataFrame(datas)
plt.bar(employees["Name"],employees["Salary"])
plt.title("employees salaries")
plt.xlabel("Employee Name")
plt.ylabel("Employee Salary")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

employee_datas = {
    "Month":["January","February","March","April","May"],
    "Sales":[12000,15000,10000,18000,22000],
    "Expenses":[8000,9000,7500,11000,14000]
}
employees = pd.DataFrame(employee_datas)
print(employees)
print("Total Sales: ",employees["Sales"].sum())
print("Average Sales: ",employees["Sales"].mean())
plt.plot(employees["Month"],employees["Sales"])
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
```

```python
employee_datas = {
    "Month":["January","February","March","April","May"],
    "Sales":[12000,15000,10000,18000,22000],
    "Expenses":[8000,9000,7500,11000,14000]
}
employees = pd.DataFrame(employee_datas)

plt.bar(employees["Month"],employees["Sales"])
plt.bar(employees["Month"],employees["Expenses"])
plt.xlabel("Month")
plt.ylabel("Sales and Expenses")
plt.title("Sales and expenses")
print("Blue is Sales and Orange is expenses")
print(employees["Sales"].max())
```
